# Retail Sales AnalysisThis notebook explores the synthetic retail sales dataset generated for the project.

```python
import pandas as pdimport matplotlib.pyplot as pltcustomers = pd.read_csv('data/customers.csv')products = pd.read_csv('data/products.csv')orders = pd.read_csv('data/orders.csv')orders['order_date'] = pd.to_datetime(orders['order_date'])
```

## Data quality and overview

```python
orders.info()orders.isna().sum()orders.describe()
```

## Sales and profit by category

```python
sales = orders.merge(products, on='product_id').merge(customers, on='customer_id')category = sales.groupby('category', as_index=False)[['sales', 'profit']].sum().sort_values('sales', ascending=False)category
```

```python
plt.style.use('seaborn-v0_8-whitegrid')fig, ax = plt.subplots(figsize=(8, 5))ax.bar(category['category'], category['sales'], color='#2563eb')ax.set_title('Sales by Product Category')ax.set_ylabel('Sales')plt.xticks(rotation=20)plt.tight_layout()plt.show()
```

## Monthly trend

```python
sales['month'] = sales['order_date'].dt.to_period('M').astype(str)monthly = sales.groupby('month', as_index=False)[['sales', 'profit']].sum()fig, ax = plt.subplots(figsize=(12, 5))ax.plot(monthly['month'], monthly['sales'], marker='o', label='Sales')ax.plot(monthly['month'], monthly['profit'], marker='o', label='Profit')ax.set_title('Monthly Sales and Profit')ax.tick_params(axis='x', rotation=75)ax.legend()plt.tight_layout()plt.show()
```
